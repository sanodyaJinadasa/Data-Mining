```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
```

```python
data={
    'weight':[1,2,2,4,5,5],
    'power':[1,2,1,4,5,4],
    'status':[0,0,0,1,1,1]
}
```

```python
df=pd.DataFrame(data)
```

```python
X=df[['weight','power']]
y=df['status']
```

```python
model=SVC(kernel='linear')
model.fit(X,y)
```

```python
model.support_vectors_
```

```python
plt.figure(figsize=(10,10))
plt.scatter(X[y==0]['weight'],X[y==0]['power'],color='red',label='Safe Gadget',marker='o',s=100)
plt.scatter(X[y==1]['weight'],X[y==1]['power'],color='blue',label='Explosive Gadget',marker='*',s=150)
```

```python
ax=plt.gca()
xlim=ax.get_xlim()
ylim=ax.get_ylim()
xx=np.linspace(xlim[0],xlim[1],30)
yy=np.linspace(ylim[0],ylim[1],30)
yy,xx=np.meshgrid(yy,xx)
xy=np.vstack([xx.ravel(),yy.ravel()]).T
z=model.decision_function(xy).reshape(xx.shape)
ax.contour(xx,yy,z,colors='k',levels=[-1,0,1],alpha=0.5,linestyles=['--','-','--'])
plt.scatter(model.support_vectors_[:,0],model.support_vectors_[:,1],s=250,linewidth=1,facecolors='none',edgecolors='k',label='Support')
plt.xlabel('Weight')
plt.ylabel('Power Level')
plt.title('Simple Linear SVM Highway')
plt.legend()
plt.grid(True)
plt.show()
```

### EXERCISE 1

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
```

```python
# 1. Create a clean, simple dataset
data = {
    'Weight': [1, 2, 2, 4, 5, 5],
    'Power': [1, 2, 1, 4, 5, 4],
    'Status': [0, 0, 0, 1, 1, 1]  # 0 = Safe (Blue), 1 = Explosive (Red)
}
df = pd.DataFrame(data)
```

```python
X = df[['Weight', 'Power']]
y = df['Status']

# 2. Initialize and Train the Linear SVM Model
# 'linear' means we want a straight decision highway
model = SVC(kernel='linear')
model.fit(X, y)
```

```python
# 3. Print out the data points acting as our Support Vectors (The Curbs)
print("--- EXPERIMENT RESULTS ---")
print("The Support Vectors are located at coordinates:")
print(model.support_vectors_)
print("--------------------------")
```

```python
# 4. Plotting the visual graph
plt.figure(figsize=(7, 5))
```

```python
# Plot the safe points as circles and explosive points as stars
plt.scatter(X[y==0]['Weight'], X[y==0]['Power'], color='blue', label='Safe Gadget', marker='o', s=100)
plt.scatter(X[y==1]['Weight'], X[y==1]['Power'], color='red', label='Explosive Gadget', marker='*', s=150)
```

```python
# Highlight the Support Vectors on the graph with green rings
# Draw the Hyperplane (Center Line) and the Margins (Gutters)
ax = plt.gca()
xlim = ax.get_xlim()
ylim = ax.get_ylim()
xx = np.linspace(xlim[0], xlim[1], 30)
yy = np.linspace(ylim[0], ylim[1], 30)
YY, XX = np.meshgrid(yy, xx)
xy = np.vstack([XX.ravel(), YY.ravel()]).T
Z = model.decision_function(xy).reshape(XX.shape)
# This draws the solid center line (0) and dashed gutters (-1, 1)
ax.contour(XX, YY, Z, colors='black', levels=[-1, 0, 1], alpha=0.5, linestyles=['--', '-', '--'])
ax.scatter(model.support_vectors_[:, 0], model.support_vectors_[:, 1], s=250, linewidth=2, facecolors='none', edgecolors='green', label='Support Vectors')
plt.xlabel('Weight')
plt.ylabel('Power Level')
plt.title('Exercise 1: Simple Linear SVM Highway')
plt.legend()
plt.grid(True)
plt.show()
```
